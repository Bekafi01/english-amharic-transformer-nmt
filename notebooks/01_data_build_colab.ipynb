# Phase 1 — Build the parallel corpus (Colab/Kaggle runner)

Thin runner: all logic lives in `src/amnmt`. This notebook only installs the package and calls the CLI.

- Raw downloads go to local disk (`/content/raw`, ~1 GB, discarded with the session).
- Outputs go to Drive: `<root>/data/processed/full/` (shards, train/valid/test parquet, `data_card.json`).
- Re-running skips sources whose shard already exists on Drive, so a session timeout costs only the current source.

```python
from google.colab import drive

drive.mount("/content/drive")
ROOT = "/content/drive/MyDrive/amnmt"  # everything persistent lives under here
```

```python
%%bash
rm -rf /content/repo
git clone --quiet https://github.com/Bekafi01/english-amharic-transformer-nmt.git /content/repo
cd /content/repo && pip install -q -e ".[data]"
amnmt version
```

```python
# Sanity run first (5k pairs per source, a few minutes). Writes to <ROOT>/data/processed/tiny
!cd /content/repo && amnmt data build -c configs/tiny.yaml --root $ROOT --raw-dir /content/raw
```

```python
# Full build. If you do not have the private `human_curated` files on Drive,
# delete that source block from configs/full.yaml before running.
!cd /content/repo && amnmt data build -c configs/full.yaml --root $ROOT --raw-dir /content/raw
```
